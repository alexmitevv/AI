<h2 style = "color: #000080;"><strong><center>Statistical Investigation Question:</strong></center></h2>

<h3 style = "color: #000080;"><center>What is the probability that a five will be rolled on a six-sided die?</center>

<h2 style = "color: #000080;"><strong><center>My Study:</strong></center></h2>

<h3 style = "color: #000080;">1. The process being simulated is rolling a fair six-sided die 1,000 times.</h3>
<h3 style = "color: #000080;">2. The trial will be a single trial - roll of the die. The possible outcomes of this trial are {1, 2, 3, 4, 5, 6}.</h3>
<h3 style = "color: #000080;">3. The event is "rolling a five." The only outcome included in this event is {5}.</h3>

<h2 style = "color: #000080;"><strong><center>My Analysis:</strong></center></h2>

```python
# "Hey computer, go grab your bag of random tricks!"
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Set seed for reproducability (optional).
np.random.seed(42)
```

```python
# 1. Simulate rolling a 6-sided die 1,000 times.
num_rolls = 1000
# "Roll a 6-sided die 1,000 times and write down every number you get!"
rolls = np.random.randint(1, 7, size = num_rolls)
```

```python
# 2. Estimate the empirical probability of rolling a 5.
# "Count how many times you got a 5!"
fives_count = np.sum(rolls == 5)
empirical_prob = fives_count / num_rolls
```

```python
# 3. Calculate the theoretical probability.
theoretical_prob = 1 / 6
```

```python
# Print results.
print("Total rolls: ", num_rolls)
print("Number of 5s rolled: ", fives_count)
print("Empirical Probability (Simulated) ", empirical_prob)
print("Theoretical Probability: ", theoretical_prob)
```

```python
# 4. Plot the outcomes.
outcomes, counts = np.unique(rolls, return_counts=True)

plt.figure(figsize=(8, 5))

# Draw the bars.
bars = plt.bar(outcomes, counts, color="skyblue", edgecolor="navy")
bars[4].set_color("dodgerblue")  # Highlight outcome '5'
bars[4].set_edgecolor("darkblue")

# Expected line.
plt.axhline(
    y=num_rolls / 6,
    color="red",
    linestyle="--",
    label="Expected Count (166.7)",
)

# Labels & formatting.
plt.title("Simulation Results: 1,000 Rolls of a 6-Sided Die", fontsize=14)
plt.xlabel("Die Outcome", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.xticks(range(1, 7))

# Fix y-axis to start at 0 so all bars are visible.
plt.ylim(0, max(counts) + 30)

plt.legend()
plt.grid(axis="y", linestyle=":", alpha=0.7)

plt.show()
```

<h2 style = "color: #000080;"><strong><center>Conclusion:</strong></center></h2>

<h3 style = "color: #000080;">1. The probability of rolling a 5 on a six-sided die is 1/6 or approximately 16.67% theoretically. In my simulation of 1,000 rolls, I rolled a 5 a total of 172 times, resulting in an empirical probability of 17.20% (0.1720).</h3>
