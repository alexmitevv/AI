<h1 style = "color: blue;"><strong><center>QUESTION:</strong></center></h1>

<h2 style = "color: gold;"><em>Out of all the individuals that eat steak, which ones consume steak AND also smoke cigarettes?</em></h2>

<h1 style = "color: blue;"><strong><center>STUDY:</strong></center></h1>

<h2 style = "color: gold;"><em>This is an observational study because I am not changing or manipulating any variables. Additionally, this study uses a sample of 551 individuals instead of a full census of the entire population.</em></h2>

<h1 style = "color: blue;"><strong><center>ANALYSIS:</strong></center></h1>

```python
import pandas as pd
```

```python
steak = pd.read_csv("9-4-2026/steak.csv")
```

```python
steak_and_smoke = steak[(steak["Do you eat steak?"] == "Yes") & (steak["Do you ever smoke cigarettes?"] == "Yes")]
```

```python
print(steak_and_smoke[["RespondentID", "Do you eat steak?", "Do you ever smoke cigarettes?"]])
```

```python
print("Total count:", len(steak_and_smoke))
```

<h1 style = "color: blue;"><strong><center>CONCLUSION:</strong></center></h1>

<h2 style = "color: gold;"><em>In summary, the frequency of my investigation is 72 individuals, signifying those that consume steak AND also smoke cigarettes. Relative frequencies: out of all steak eaters (427 total), approximmately 16.86% of steak eaters smoke cigarettes, whereas the other 83.14% of steak eaters don't smoke cigarettes. The mode is "No" because there are more steak eaters that are non-smokers compared to the smokers.
WORK SHOWN BELOW:</em></h2>

```python
steak_eaters = steak[steak["Do you eat steak?"] == "Yes"]
```

<h1 style = "color: blue;"><strong><center>FREQUENCY:</strong></center></h1>

```python
smoking_counts = steak_eaters["Do you ever smoke cigarettes?"].value_counts()
print("=== Counts ===")
print(smoking_counts)
```

<h1 style = "color: blue;"><strong><center>RELATIVE FREQUENCY (PERCENTAGES):</strong></center></h1>

```python
smoking_rel_freq = (steak_eaters["Do you ever smoke cigarettes?"].value_counts(normalize = True) * 100)
print("\n=== Relative Frequencies (%) ===")
print(smoking_rel_freq.round(2))
```

<h1 style = "color: blue;"><strong><center>MODE:</strong><center></h1>

```python
smoking_mode = steak_eaters["Do you ever smoke cigarettes?"].mode()[0]
```

```python
print("\n=== Mode ===")
print(smoking_mode)
```
