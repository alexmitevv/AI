```python
import pandas as pd
import seaborn as sns
```

```python
mushrooms = pd.read_csv("mushrooms.csv")
```

```python
mushrooms.head()
```

```python
odor_labels = {'a': 'Almond', '1': 'Anise', 'c': 'Creosote', 'y': 'fishy', 'f': 'foul', 'm': 'musty', 'n': 'none', 'p': 'pungent', 's': 'spicy'}
mushrooms['odor_labels'] = mushrooms['odor'].map(odor_labels)
```

```python
poisonous = mushrooms[mushrooms['class'] == 'p']
```

```python
odor_counts = poisonous['odor_labels'].value_counts()
```
