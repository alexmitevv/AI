<h3 style="color: lightblue;"><em><center>Which habitat is the most common?</center></em></h3>

<h2 style="color: blue;"><strong><center>Study Description:</center></strong></h2>

<h3 style="color: lightblue;"><em><center>Observational study because I am only analyzing a sample of mushrooms from across the globe, though I am examining the entire dataset.</center></em></h3>

```python
import pandas as pd
import os
```

```python
mushrooms = pd.read_csv("mushrooms.csv")
```

```python
habitat_labels = {'g': 'grasses', 'l': 'leaves', 'm': 'meadows', 'p': 'paths', 'u': 'urban', 'w': 'waste', 'd': 'woods'}
```

```python
mushrooms['habitat_name'] = mushrooms['habitat'].map(habitat_labels)
```

```python
frequency = mushrooms['habitat_name'].value_counts()
```

```python
relative_frequency = mushrooms['habitat_name'].value_counts(normalize = True)
```

```python
habitat_mode = mushrooms['habitat_name'].mode()[0]
```

```python
summary_df = pd.DataFrame({'Frequency': frequency, 'Relative Frequency': relative_frequency})
```

```python
print(summary_df)
```

```python
print(f"\nThe mode (most commmon habitat) is {habitat_mode}")
```
