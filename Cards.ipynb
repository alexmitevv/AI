<h1 style = "color: #000080;"><strong><em><center>Cards Simulation Study - Joint Probability</strong></em></center></h1>

<h2 style = "color: #000080;"><strong><center>Statistical Investigation Question:</strong></center></h2>

<h3 style = "color: #000080;"><center>In a game of Go Fish, if you draw 2 cards in a row from a full standard 52-card deck without replacing them, what is the probability of drawing a King on your first draw AND an Ace on your second draw?</center>

<h2 style = "color: #000080;"><strong><center>Simulation's Characteristics:</strong></center></h2>

<h3 style = "color: #000080;">1. The random process the act of randomly drawing two cards - one after the other.

<h3 style = "color: #000080;">2. One trial worth performing when two cards are drawn. The possible outcomes are any two cards found in a 52-card deck.

<h3 style = "color: #000080;">3. The event I'll be measuring is the act of drawing a King followed by an Ace.

<h2 style = "color: #000080;"><strong><center>Simulation's Revised Characteristics:</strong></center></h2>

<h3 style = "color: #000080;">1. A = drawing a King on 1st draw, B = drawing an Ace on 2nd draw. In other words, I will call the event of drawing a King on the first draw "A" and the event of drawing an Ace on the second draw "B".

<h3 style = "color: #000080;">2. My question is asking P(A⋂B), also known as a successful example of joint probability.

<h3 style = "color: #000080;">3. A and B are NOT mutually exclusive because drawing a King doesn't make it impossible to draw an Ace next. Drawing a King on the first draw doesn't prevent me from getting an Ace on the second draw.

<h3 style = "color: #000080;">4. A and B aren't independent, which means they're dependent on one another. This is because rolling a card on the first draw doesn't impact what will be drawn on the second draw. Drawing a King for the first draw changes the probability of drawing an Ace on the second draw.

```python
# Import necessary libraries.
import random
import matplotlib.pyplot as plt
```

```python
# 1. Set up variables.
num_trials = 1000
successes = 0
```

```python
# 2. Build a standard deck of 52 cards (4 Kings, 4 Aces, 44 others).
deck = ["K"] * 4 + ["A"] * 4 + ["Other"] * 44
```

```python
# 3. Run simulation 1,000 times.
for _ in range(num_trials):
    # Draw 2 cards without replacement.
    hand = random.sample(deck, 2)
    # Check if the 1st card is a King AND the 2nd card is an Ace.
    if hand[0] == "K" and hand[1] == "A":
        successes= successes + 1
```

```python
# 4. Calculate probabilities.
simulated_prob = successes / num_trials
theoretical_prob = (4 / 52) * (4/51)
```

```python
# 5. Display results.
print("Total Trials: " + str(num_trials))
print("Successful Draws: " + str(successes))
print("Simulated Probability: " + str(round(simulated_prob, 5)))
print("Theoretical Probability: " + str(round(theoretical_prob, 5)))
```

```python
# 6. Empirical & Theoretical Calculations:
empirical_prob = successes / num_trials
theoretrical_prob = (4 / 52) * (4 / 51)
print("Empirical Probability: " + str(round(empirical_prob, 5)))
print("Theoretical Probability: " + str(round(theoretical_prob, 5)))
```

```python
# 7. Graphing results:
categories = ["Empirical\n(Simulated)", "Theoretical\n(Exact)"]
probabilities = [empirical_prob, theoretical_prob]
```

```python
# 8. Create bar chart.
bars = plt.bar(
    categories, probabilities, color=["skyblue", "salmon"], edgecolor="black"
)

# 9. Add exact numbers above each bar.
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2.0,
        height + 0.0001,
        str(round(height, 5)),
        ha="center",
        va="bottom",
    )

# 10. Add titles, labels, & grid.
plt.title("Go Fish: Drawing a King then an Ace (1,000 Trials)")
plt.ylabel("Probability")
plt.ylim(0, max(probabilities) * 1.3)
plt.grid(axis="y", linestyle="--", alpha=0.7)

# Display completed plot.
plt.show()
```

<h3 style = "color: #000080;"><center>In a game of Go Fish, if you draw 2 cards in a row from a full standard 52-card deck without replacing them, the probability of drawing a King on your first draw AND an Ace on your second draw is 0.007 empirically and 0.00603 theoretically.
