<h1><strong> Analysis: </strong></h1>

```python
import pandas as pd
```

```python
titanic = pd.read_csv("titanic.csv")
```

```python
male_count = (titanic['Sex'] == "male").sum()
```

```python
print(male_count)
```

```python
males_count = (titanic[titanic["Sex"] == "male"])
```

```python
count = len(males_count)
print("Number of males that were on the Titanic:", count)
```
